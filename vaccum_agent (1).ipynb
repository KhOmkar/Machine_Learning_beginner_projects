# Vacuum Cleaner Agent Simulation

This notebook simulates a **vacuum cleaner agent** operating in a simple environment with multiple rooms. Each room can be either **Clean** or **Dirty**, and the agent’s goal is to clean all dirty rooms.

## Key Features

1. **Random Environment Initialization**  
   - Each room is randomly set as clean (`C`) or dirty (`D`) at the start.  
   - The agent starts in a random room.

2. **Two-Row Terminal Visualization**  
   - **First row** shows room states: `[C]` = Clean, `[D]` = Dirty.  
   - **Second row** shows agent position with `A` under the current room.  
   - Each step displays the **action taken** and **current performance score**.

3. **Agent Behavior**  
   - If the current room is **dirty**, the agent **cleans it**.  
   - If the room is clean, the agent **moves** left or right.  
   - Invalid moves (moving outside the environment) are **avoided**.

4. **Performance Score**  
   - +5 points for cleaning a dirty room.  
   - -2 points for cleaning an already clean room.  
   - -1 point for moving.  
   - -2 points for attempting an invalid move (though prevented by the agent).

5. **Stopping Condition**  
   - The simulation **stops automatically** when all rooms are clean.  
   - The total number of steps and final performance score are displayed at the end.

## How It Works

- The **Environment class** handles room states, agent location, cleaning, moving, and visualization.  
- The **Agent class** decides actions based on the current room state and valid moves.  
- The **Simulation loop** repeatedly lets the agent act and updates the display until all rooms are clean.


```python
import random
import time
import os

# Environment class representing the vacuum cleaner world
class VacuumEnvironment:
    def __init__(self, num_rooms=3):
        self.num_rooms = num_rooms
        self.rooms = {i: random.choice(['Clean', 'Dirty']) for i in range(num_rooms)}
        self.agent_location = random.randint(0, num_rooms-1)
        self.performance_score = 0

    # Check if a specific room is dirty
    def is_dirty(self, room):
        return self.rooms[room] == 'Dirty'

    # Clean the current room and update performance
    def clean(self, room):
        if self.is_dirty(room):
            self.rooms[room] = 'Clean'
            self.performance_score += 5   # Reward for cleaning dirty room
        else:
            self.performance_score -= 2    # Penalty for cleaning already clean room

    # Move the agent left or right and update performance
    def move(self, direction):
        if direction == "LEFT" and self.agent_location > 0:
            self.agent_location -= 1
            self.performance_score -= 1
        elif direction == "RIGHT" and self.agent_location < self.num_rooms-1:
            self.agent_location += 1
            self.performance_score -= 1
        else:
            self.performance_score -= 2  # Penalty for invalid move

    # Check if all rooms are clean
    def all_clean(self):
        return all(state == 'Clean' for state in self.rooms.values())

    # Display the environment in terminal with two rows
    def display(self, action=""):
        os.system('cls' if os.name == 'nt' else 'clear')
        room_row = ""
        agent_row = ""
        for i in range(self.num_rooms):
            room_row += "[D]" if self.rooms[i] == 'Dirty' else "[C]"
            agent_row += " A " if i == self.agent_location else "   "
        print(room_row)
        print(agent_row)
        if action:
            print("Action Taken:", action)
        print("Performance Score:", self.performance_score)
        print("-" * (self.num_rooms * 3))


# Agent class that decides the actions of the vacuum
class VacuumAgent:
    def __init__(self, environment):
        self.env = environment

    # Decide and perform an action based on current room state
    def act(self):
        current_room = self.env.agent_location
        if self.env.is_dirty(current_room):
            # If room is dirty, clean it
            self.env.clean(current_room)
            return "CLEAN"
        else:
            # Determine valid moves (avoid invalid moves)
            valid_moves = []
            if self.env.agent_location > 0:
                valid_moves.append("LEFT")
            if self.env.agent_location < self.env.num_rooms - 1:
                valid_moves.append("RIGHT")

            # Choose a valid move randomly
            move = random.choice(valid_moves)
            self.env.move(move)
            return f"MOVE {move}"


# Simulation
if __name__ == "__main__":
    env = VacuumEnvironment(num_rooms=3)
    agent = VacuumAgent(env)

    # Display initial state
    print("Initial State (Rooms: C=Clean, D=Dirty, Agent=A):")
    env.display("START")
    time.sleep(2)

    step = 0
    while not env.all_clean():
        step += 1
        action = agent.act()
        env.display(action)
        time.sleep(1)

    print(f"\nAll rooms are clean after {step} steps!")
    print("Final Performance Score:", env.performance_score)
    print("Final Room States:", env.rooms)
```
